```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
combine_results = pd.read_csv("data/combine_results.csv")
combine_tracking = pd.read_csv("data/combine_tracking.csv")
career_successes = pd.read_csv("data/player_career_successes.csv")
player_play = pd.read_csv("data/player_play.csv")

print("All four datasets loaded successfully.")
```

```python
datasets = {
    "Combine Results": combine_results,
    "Combine Tracking": combine_tracking,
    "Career Successes": career_successes,
    "Player Play": player_play
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]:,}")
    print("Column names:")
    print(df.columns.tolist())
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print(f"Unique players: {df['nfl_id'].nunique():,}")
    print(f"Missing nfl_id values: {df['nfl_id'].isna().sum():,}")
    print(f"Total missing values: {df.isna().sum().sum():,}")

    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    if not missing.empty:
        print("\nColumns with missing values:")
        print(missing)
    else:
        print("No missing values.")
```

```python
combine_ids = set(combine_results["nfl_id"])
tracking_ids = set(combine_tracking["nfl_id"])
career_ids = set(career_successes["nfl_id"])
play_ids = set(player_play["nfl_id"])

print("Combine and Tracking match:",
      len(combine_ids & tracking_ids), "/",
      len(combine_ids))

print("Combine and Career match:",
      len(combine_ids & career_ids), "/",
      len(combine_ids))

print("Combine and Player Play match:",
      len(combine_ids & play_ids), "/",
      len(combine_ids))

print("\nPlayers missing from Player Play:")
print(sorted(combine_ids - play_ids))
```

```python
print("Drill types:")
print(combine_tracking["drill_type"].value_counts(dropna=False))

print("\nDrill names:")
print(combine_tracking["drill_name"].value_counts(dropna=False))
```

```python
standard_drills = [
    "FORTY_YARD_DASH",
    "THREE_CONE_DRILL",
    "SHORT_SHUTTLE"
]

drill_attempts = (
    combine_tracking[
        combine_tracking["drill_name"].isin(standard_drills)
    ]
    .groupby(["drill_name", "nfl_id"])["attempt"]
    .nunique()
    .reset_index(name="attempt_count")
)

print(drill_attempts.groupby("drill_name")["attempt_count"].describe())
```

```python
for drill, result_column in [
    ("FORTY_YARD_DASH", "forty"),
    ("SHORT_SHUTTLE", "short_shuttle"),
    ("THREE_CONE_DRILL", "three_cone")
]:
    tracking_players = set(
        combine_tracking.loc[
            combine_tracking["drill_name"] == drill, "nfl_id"
        ]
    )

    result_players = set(
        combine_results.loc[
            combine_results[result_column].notna(), "nfl_id"
        ]
    )

    print(f"\n{drill}")
    print("Players with tracking:", len(tracking_players))
    print("Players with traditional result:", len(result_players))
    print("Players with both:", len(tracking_players & result_players))
```
